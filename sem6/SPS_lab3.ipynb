<h1> Spatial </h1>

```python
ts1 = np.array(ts1)
ts2 = np.array(ts2)

N = len(ts1)
fs = sample_rate

freqs = np.fft.fftfreq(N, d = 1/fs)

fft_ts1 = np.fft.fft(ts1)
fft_ts2 = np.fft.fft(ts2)
```

```python
# using only positive frequencies

half_N = N // 2

plt.figure(figsize=(12, 4))
plt.plot(freqs[:half_N], np.abs(fft_ts1[:half_N]))
plt.title("Frequency Domain Signal: ts1 (697 Hz + 1209 Hz)")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

plt.figure(figsize = (12, 4))
plt.plot(freqs[:half_N], np.abs(fft_ts2[:half_N]))
plt.title("Frequence Domain Signal: ts2 (697 Hz + 1336 Hz)")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()


```

```python
ts = np.array(ts)
fs = sample_rate
plt.figure(figsize=(12, 4))

plt.specgram(ts, NFFT = 1024, Fs = fs, noverlap = 512, cmap = "jet")

plt.title("Spectogram of Signal (ts)")
plt.xlabel("Time (seconds)")
plt.ylabel("Frequency (Hz)")
plt.colorbar(label = "Intensity (Db)")
plt.show()
```
